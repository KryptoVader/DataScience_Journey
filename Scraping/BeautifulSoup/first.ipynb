    .decompose() removes a tag from the tree, then completely destroys it and its contents:

```python
markup = '<a href="http://example.com/">I linked to <i>example.com</i></a>'
anchor = BeautifulSoup(markup)
a_tag = anchor.a

anchor.i.decompose()

print(a_tag)
print(anchor) # As you can see that the i tag along with its content is removed and destroyed from the HTML Tree and even in the anchor tag also which was stored in a_tag
```

    The .replace_with() is used to replace a particular tag with another tag

```python
markup = '<a href="http://example.com/">I linked to <i>example.com</i></a>'
mar = BeautifulSoup(markup)
a_tag = mar.a

new_tag = mar.new_tag("b")
new_tag.string = "example.net"
a_tag.i.replace_with(new_tag)

a_tag # The i tag is replaced with the b tag that we created with help of new_tag()
```

    The wrap() and unwrap() is used to wrap and unwrap the parent about a particular element

```python
markup = '<p>text</p>'
mar = BeautifulSoup(markup)

div = mar.new_tag('div')
mar.p.wrap(div)
print(mar)
```

```python
mar.p.unwrap()
mar
```

# We use SoupStrainer to parse only particular elements we want instead of the whole document

```python
from bs4 import SoupStrainer
a_tags = SoupStrainer('a')

Soup = BeautifulSoup(requests.get(URL).text, parse_only=a_tags)
print(Soup.prettify())
```
