```python
from sklearn.model_selection import train_test_split
X = plants.drop(columns=['Plant Type'])
y = plants['Plant Type']                  
# Now split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
X_train
```

## Standardization

` Xi' = (Xi - mean)/std dev `

```python
X_train_scaled_manual = pd.DataFrame()
for i in X_train.columns:
    X_train_scaled_manual[i] = (X_train[i] - X_train[i].mean()) / X_train[i].std()

X_train_scaled_manual
```

```python
sns.boxplot(data = X_train_scaled_manual, x = 'pH')
plt.show()
```

## Normalization

1) MinMax
2) Mean
3) Max Absolute
4) Robust

### MinMax 

` Xi' = (X - Xmin)/ (Xmax - Xmin) `

```python
X_train_scaled_manual = pd.DataFrame()
for i in X_train.columns:
    X_train_scaled_manual[i] = (X_train[i] - X_train[i].min()) / (X_train[i].max() - X_train[i].min())

X_train_scaled_manual
```

```python
sns.boxplot(data = X_train_scaled_manual, x = 'pH')
plt.show()
```

```python
from sklearn.preprocessing import MinMaxScaler

minMaxScaler = MinMaxScaler()
X_train_scaled = pd.DataFrame(minMaxScaler.fit_transform(X_train), columns = X_train.columns)
X_train_scaled
```

```python
sns.boxplot(data = X_train_scaled, x = 'pH')
plt.show()
```

### Mean Scaler

` Xi' = (Xi - mean(Xi)) / (max(Xi) - min(Xi))`

```python
X_train_scaled_manual = pd.DataFrame()
for i in X_train.columns:
    X_train_scaled_manual[i] = (X_train[i] - X_train[i].mean()) / (X_train[i].max() - X_train[i].min())

X_train_scaled_manual
```

```python
sns.boxplot(data = X_train_scaled_manual, x = 'pH')
plt.show()
```

### Max Absolute 

` Xi' = Xi / abs(max(Xi)) `

```python
X_train_scaled_manual = X_train / X_train.abs().max()
X_train_scaled_manual
```

```python
sns.boxplot(data = X_train_scaled_manual, x = 'pH')
plt.show()
```

```python
from sklearn.preprocessing import MaxAbsScaler
maxAbsScale = MaxAbsScaler()
X_train_scaled = pd.DataFrame(maxAbsScale.fit_transform(X_train), columns = X_train.columns)
X_train_scaled
```

```python
sns.boxplot(data = X_train_scaled, x = 'pH')
plt.show()
```

### Robust Scaler

` Xi' = (Xi - Q2) / (Q3 - Q1) `

```python
X_train_scaled_manual = (X_train - X_train.quantile(0.50)) / (X_train.quantile(0.75) - X_train.quantile(0.25))
X_train_scaled_manual
```

```python
sns.boxplot(data = X_train_scaled_manual, x = 'pH')
plt.show()
```

#### Insights 

The Graph generated by Robust Scaler has made sure that outlier doesnt affect the median like see the median is at 0 while in Standard Classifier the median was below 0
